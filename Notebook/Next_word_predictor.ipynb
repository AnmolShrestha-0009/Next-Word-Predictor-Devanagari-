```python
def word_tokenize(sentence):
    if isinstance(sentence, list):
        sentence = ' '.join(sentence)
    if not isinstance(sentence, str):
        return []
    tokens = re.findall(r'[\u0900-\u097F]+|[a-zA-Z]+|\d+|[^\s\u0900-\u097Fa-zA-Z\d]', sentence)
    return tokens

df['concate_text'] = df['concate_text'].apply(word_tokenize)
```

## Fine-Tuning NepaliBERT on the Corpus + Semantic Similarity Search

`category` was dropped earlier, so we don't have labels for supervised fine-tuning.
Instead we do **domain-adaptive fine-tuning via Masked Language Modeling (MLM)**:
this continues pretraining NepaliBERT on your own articles so its representations
better fit your corpus's vocabulary and style, without needing labels.


```python
clean_text=[]
for items in df['concate_text']:
    if isinstance(items,list):
      text_string=" ".join(items)
    else:
      text_string=str(items)
    clean_text.append(text_string)
df["bert_model"]=clean_text

```

```python
import torch
from torch.utils.data import Dataset
from transformers import (
    AutoTokenizer, AutoModelForMaskedLM,
    DataCollatorForLanguageModeling, TrainingArguments, Trainer
)

MODEL_NAME = "Rajan/NepaliBERT"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
mlm_model = AutoModelForMaskedLM.from_pretrained(MODEL_NAME)
```

```python
SAMPLE_SIZE =  len(df)
texts = df['bert_model'].sample(SAMPLE_SIZE, random_state=42).tolist()

class MLMDataset(Dataset):
    def __init__(self, texts, tokenizer,max_len=1000):
        self.texts = texts
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        # Tokenize dynamically per item without forcing return_tensors='pt' here
        # The DataCollator will handle tensor conversion and padding automatically
        encoding = self.tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=self.max_len,
            padding='max_length' # Optional: DataCollator can also handle padding dynamically
        )
        return encoding

# Initialize dataset and collator
train_dataset = MLMDataset(texts, tokenizer)

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=True,
    mlm_probability=0.15
)
```

```python
training_args = TrainingArguments(
    output_dir="./nepalibert-finetuned",
    num_train_epochs=2,
    per_device_train_batch_size=8,
    save_strategy="epoch",
    logging_steps=50,
    learning_rate=5e-5,
    fp16=torch.cuda.is_available(),
    report_to="none",
)

trainer = Trainer(
    model=mlm_model,
    args=training_args,
    train_dataset=train_dataset,
    data_collator=data_collator,
)

# trainer.train()
trainer.save_model("./nepalibert-finetuned")
tokenizer.save_pretrained("./nepalibert-finetuned")
```

### Semantic Similarity Search using the Fine-Tuned Model

We embed each document by mean-pooling the fine-tuned model's token embeddings
(masking out padding), then use cosine similarity to find the most similar
articles to a query.

```python
from transformers import AutoModel
import torch.nn.functional as F
import numpy as np

finetuned_model = AutoModel.from_pretrained("./nepalibert-finetuned")
finetuned_model.eval()

def get_embedding(text, model=finetuned_model, tokenizer=tokenizer, max_len=128):
    inputs = tokenizer(text, truncation=True, padding=True,
                        max_length=max_len, return_tensors='pt')
    with torch.no_grad():
        outputs = model(**inputs)
    token_embeds = outputs.last_hidden_state
    mask = inputs['attention_mask'].unsqueeze(-1).expand(token_embeds.size()).float()
    summed = torch.sum(token_embeds * mask, dim=1)
    counts = torch.clamp(mask.sum(dim=1), min=1e-9)
    mean_pooled = summed / counts
    return F.normalize(mean_pooled, p=2, dim=1).squeeze().numpy()
```

```python
corpus_sample = df["bert_model"].sample(min(1000, len(df)), random_state=42).reset_index(drop=True)
corpus_embeddings = np.vstack([get_embedding(t) for t in corpus_sample])

def most_similar(query, top_n=5):
    query_emb = get_embedding(query)
    sims = corpus_embeddings @ query_emb
    top_idx = np.argsort(-sims)[:top_n]
    for rank, idx in enumerate(top_idx, 1):
        print(f"{rank}. (score={sims[idx]:.3f}) {corpus_sample.iloc[idx][:120]}...")
    return top_idx, sims[top_idx]

most_similar("नेपालको अर्थतन्त्र र बजेट", top_n=5)
```


## Next-Word Predictor: Pretrained NepaliBERT Embeddings + LSTM


```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from collections import Counter

# df['concate_text'] currently holds lists of word tokens (from word_tokenize() above)
all_tokens = [tok for doc in df['concate_text'] for tok in doc]
print("total tokens:", len(all_tokens))

VOCAB_SIZE = 20000  # cap vocab to the most frequent words; raise if you have the memory
counts = Counter(all_tokens)
most_common = counts.most_common(VOCAB_SIZE - 2)  # reserve 2 slots for special tokens

PAD, UNK = "<pad>", "<unk>"
idx2word = [PAD, UNK] + [w for w, _ in most_common]
word2idx = {w: i for i, w in enumerate(idx2word)}
print("vocab size:", len(idx2word))
```

```python

bert_embedding_layer = model.get_input_embeddings()  # model = NepaliBERT loaded earlier
EMBED_DIM = bert_embedding_layer.embedding_dim

embedding_matrix = torch.zeros(len(idx2word), EMBED_DIM)

with torch.no_grad():
    for i, w in enumerate(idx2word):
        if w in (PAD, UNK):
            continue  # leave as zeros / let the model learn them
        sub_ids = tokenizer(w, add_special_tokens=False)["input_ids"]
        if not sub_ids:
            continue
        sub_ids = torch.tensor(sub_ids)
        embedding_matrix[i] = bert_embedding_layer(sub_ids).mean(dim=0)

print(embedding_matrix.shape)
```

```python


from torch.utils.data import Dataset, DataLoader

CONTEXT_SIZE = 4  
def encode(tok):
    return word2idx.get(tok, word2idx[UNK])

sequences = []
for doc in df['concate_text']:
    ids = [encode(t) for t in doc]
    for i in range(CONTEXT_SIZE, len(ids)):
        context = ids[i - CONTEXT_SIZE:i]
        target = ids[i]
        sequences.append((context, target))

print("training pairs:", len(sequences))

class NextWordDataset(Dataset):
    def __init__(self, sequences):
        self.sequences = sequences
    def __len__(self):
        return len(self.sequences)
    def __getitem__(self, idx):
        context, target = self.sequences[idx]
        return torch.tensor(context, dtype=torch.long), torch.tensor(target, dtype=torch.long)

train_loader = DataLoader(NextWordDataset(sequences), batch_size=64, shuffle=True)
```

```python


class NextWordLSTM(nn.Module):
    def __init__(self, embedding_matrix, hidden_dim=256, num_layers=1, freeze_embeddings=False):
        super().__init__()
        vocab_size, embed_dim = embedding_matrix.shape
        self.embedding = nn.Embedding.from_pretrained(embedding_matrix, freeze=freeze_embeddings)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, num_layers=num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_dim, vocab_size)
        self.bn = nn.BatchNorm1d(hidden_dim)
        self.bn = nn.BatchNorm1d(hidden_dim)

    def forward(self, x):
        embedded = self.embedding(x)          # (batch, seq_len, embed_dim)
        output, _ = self.lstm(embedded)        # (batch, seq_len, hidden_dim)
        last_hidden = output[:, -1, :]         # (batch, hidden_dim)
        logits = self.fc(last_hidden)          # (batch, vocab_size)
        return logits

device = "cuda" if torch.cuda.is_available() else "cpu"
lstm_model = NextWordLSTM(embedding_matrix).to(device)
```

```python
import copy
class EarlyStopping():
  def __init__(self,patience,min_delta=0.0):
        self.patience = patience
        self.min_delta = min_delta
        self.counter = 0
        self.best_loss = None
        self.early_stop = False
        self.best_model_wts = None
  def __call__(self, val_loss, model):
        if self.best_loss is None:
            self.best_loss = val_loss
            self.best_model_wts = copy.deepcopy(model.state_dict())
        elif val_loss > self.best_loss - self.min_delta:
            self.counter += 1
            print(f"EarlyStopping tracker: {self.counter} / {self.patience} epochs without improvement.")
            if self.counter >= self.patience:
                self.early_stop = True
        else:
            self.best_loss = val_loss
            self.best_model_wts = copy.deepcopy(model.state_dict())
            self.counter = 0
```

```python
optimizer = torch.optim.Adam(lstm_model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

EPOCHS = 3
lstm_model.train()
for epoch in range(EPOCHS):
    total_loss = 0.0
    for context, target in train_loader:
        context, target = context.to(device), target.to(device)
        optimizer.zero_grad()
        logits = lstm_model(context)
        early_stopping = EarlyStopping(patience=3, min_delta=0.001)
        loss = criterion(logits, target)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
    print(f"epoch {epoch+1}/{EPOCHS} - avg loss: {total_loss/len(train_loader):.4f}")
```

```python
# Top-N next-word prediction with probabilities.

def predict_next_word(text, top_n=5):
    tokens = word_tokenize(text)[-CONTEXT_SIZE:]  
    ids = [encode(t) for t in tokens]
    ids = [word2idx[PAD]] * (CONTEXT_SIZE - len(ids)) + ids  # left-pad if too short
    x = torch.tensor([ids], dtype=torch.long).to(device)

    lstm_model.eval()
    with torch.no_grad():
        logits = lstm_model(x)
        probs = F.softmax(logits, dim=-1).squeeze(0)

    top_probs, top_idx = torch.topk(probs, top_n)
    print(f"Context: {' '.join(tokens)}")
    for rank, (p, idx) in enumerate(zip(top_probs, top_idx), 1):
        print(f"  {rank}. {idx2word[idx.item()]:<15} p={p.item():.4f}")
    return [(idx2word[idx.item()], p.item()) for p, idx in zip(top_probs, top_idx)]

predict_next_word("नेपाल सरकारले आज", top_n=5)
```
